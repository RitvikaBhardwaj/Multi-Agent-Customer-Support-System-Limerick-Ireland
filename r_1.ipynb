```python
%%R

# Create student data
names <- c("Aman", "Riya", "Kabira", "Snehal", "Arjun")
marks <- c(78, 85, 92, 74, 88)

# Create dataframe
students <- data.frame(names, marks)

# Print dataframe
print(students)

# Average marks
avg_marks <- mean(marks)
print(paste("Average Marks:", avg_marks))

# Highest marks
highest <- max(marks)
print(paste("Highest Marks:", highest))

# Lowest marks
lowest <- min(marks)
print(paste("Lowest Marks:", lowest))

# Plot marks
barplot(marks, names.arg = names,
        main = "Student Marks",
        xlab = "Students",
        ylab = "Marks")
```
